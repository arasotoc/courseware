# Git, GitHub, and a Reliable Course Workflow

This reading explains how your files move among GitHub, Colab, and a full computing workspace. The goal is a small workflow you can understand, repeat, and recover when something goes wrong.

By the end, you should be able to:

- distinguish Git, GitHub, a repository, and a commit;
- make your own fork of a course repository and open it in Colab;
- save meaningful checkpoints and inspect their history;
- use the minimum command-line Git loop in a full workspace; and
- recognize safe recovery and authentication practices.

Current course repositories are published through the [DS 3001 course website](https://github.com/sds3001).

## Four terms

- **Git** is version-control software. It records selected checkpoints of a project and connects those checkpoints into a history.
- **GitHub** is an online service that hosts Git repositories. Git and GitHub are related, but they are not the same thing.
- A **repository** (or **repo**) is a project folder whose files and committed history Git manages.
- A **commit** is a named checkpoint: conceptually, a snapshot of the tracked files at a useful moment.

Two more terms help once you use Git at the command line. Your **working tree** is the set of files you are editing now. The **staging area** is the proposed contents of your next commit. Git stores commits efficiently, so a repository can preserve a long history without keeping a separate visible copy of the folder for every checkpoint.

## Separate the durable copy from the working environment

Your GitHub repository is the durable online copy you can reach from different environments. Colab, a laptop, or a hosted computing service is where Python actually runs. A notebook open in a temporary environment is not automatically safe just because the repository exists online: the useful changes become durable only after you save and commit them.

A **remote** is a repository elsewhere—usually on GitHub—that your local repository knows how to contact. `push` sends your commits to a remote; `pull` brings remote commits into your current local branch.

## Choose the environment

| Start with Colab | Use a full local or hosted workspace when needed |
|---|---|
| Little or no local setup | The whole repository is available at once |
| Convenient for one notebook | Files can persist between sessions |
| GitHub actions are available through menus | Git commands and editor integration are available |
| Enough compute for much early course work | More storage, memory, or compute may be available |

The notebook is still a Jupyter notebook in either setting. Use Colab, JupyterLab, VS Code, or another suitable editor; what matters is that the work you want to keep is committed to the correct repository.

## Before you begin

Create a GitHub account if you do not already have one, then sign in. Open the course website at <https://github.com/sds3001> and choose the repository named by the instructor. A repository may contain notebooks, code, data, and documentation together.

Do not place passwords, access tokens, private keys, or other secrets in a repository or notebook. A private repository limits who can browse it, but it is not a safe secret store.

## Fork once: make the course repository yours

A **fork** is a GitHub-hosted copy of a repository under another account. Open the assigned course repository and choose **Fork**.

![The Fork control on a GitHub repository page.](./src/fork.png)

After GitHub finishes, check the owner shown near the repository name. It should be your GitHub username, with a note that the repository was forked from the course copy. Work in your fork so your commits belong to your repository rather than the course original.

## Open your fork in Colab

Colab can load a notebook directly from GitHub. In Colab, choose **File → Open notebook**, select the **GitHub** source, and search for your username or paste the URL of your fork.

![Colab's notebook picker with the GitHub source selected.](./src/colab.png)

Before editing, make one more owner check: the selected notebook should come from your account. This avoids doing substantial work in a read-only course copy or an unrelated repository.

## Save a meaningful checkpoint

When the notebook reaches a useful state, choose **File → Save a copy in GitHub**. Select your fork, confirm the file path, and describe the change in the commit message.

![Colab's form for saving a notebook to GitHub with a commit message.](./src/commit.png)

A helpful commit message is short, specific, and written in the present tense, such as `Add data-cleaning checks`. `Update notebook` says very little; `Add missing-value checks before modeling` will help you and your collaborators later.

## Find the commit history

Return to your fork on GitHub and follow the commits link near the current branch.

![A repository page with the commit-history link highlighted.](./src/commit_history_1.png)

The history answers three practical questions: *What was saved? When was it saved? What message explains the change?*

## Read the commit history

![A GitHub commit-history page showing a sequence of saved changes.](./src/commit_history_2.png)

Each row is a checkpoint. Open a commit to inspect which files changed and the associated **diff**, a line-by-line description of changes. Git conceptually records snapshots while using differences and other internal structures to store them efficiently. Only committed states belong to this history; unsaved edits in a temporary Colab session do not.

## The Colab loop

The workflow to remember is:

**fork once → open your fork → edit → save a commit → check the history**

Commit whenever you reach a useful checkpoint—not after every keystroke, but often enough that the message describes one coherent step. Because a temporary Colab runtime can disappear, confirm important work on GitHub before ending the session.

## The same ideas in a full workspace

A full workspace usually gives you a terminal and the entire repository. You will use Git commands instead of Colab's save dialog. The commands below are examples to read; this notebook contains no executable shell cells and will not change your files.

A terminal command runs in a shell. In Jupyter, a leading `!` sends one line to the shell instead of the Python interpreter—for example, `!pwd` asks the shell for the present working directory. In a terminal, omit the `!`. Always check which directory you are in before running a Git command.

## Clone the whole repository once

**Question:** How do you bring every tracked file and the committed history into a new workspace?

```bash
git clone https://github.com/<username>/<repo>.git
cd <repo>
```

Replace `<username>` and `<repo>` with the owner and repository name of **your fork**. `clone` creates a new local folder, downloads the committed project history, and configures the GitHub repository as a remote. Clone into a location where a folder with that name does not already exist. For an existing clone, use `pull`; do not keep cloning over the same working folder.

## Inspect, stage, commit, and share

**Question:** Which changes belong in the next checkpoint?

```bash
git status
git diff
git add <file>
git diff --staged
git commit -m "Add data-cleaning checks"
git push
```

Read this as a sequence. `status` summarizes the working tree. `diff` shows unstaged edits. `add` puts the named file in the staging area. `diff --staged` previews the proposed checkpoint. `commit` records that staged checkpoint locally. `push` sends new commits to the configured GitHub remote.

Prefer explicit `git add <file>` over a shortcut that commits every tracked edit. The explicit step makes it easier to notice an unrelated change before it enters the history.

## Git tracks intentional changes

A newly created file is **untracked** until you stage and commit it. If a notebook or data file seems to be missing from GitHub, start with:

```bash
git status
git ls-files
```

`git status` reports untracked, modified, and staged files. `git ls-files` lists paths Git already tracks. Use `git add <file>` to stage a new file. Use `git rm <file>` only when you intend to remove the file from both the working tree and the next commit; inspect `git status` before committing the deletion.

## Pull before local and remote work diverge

**Question:** What should you do when GitHub has commits that your local copy does not?

```bash
git status
git pull
```

First make sure you understand any local changes. Then `pull` retrieves remote commits and integrates them into the current branch. Afterward, return to the usual loop: edit → status → add → commit → push. If Git reports a conflict, pause and read the named files instead of repeatedly running commands or forcing a push.

## Authenticate without exposing a secret

GitHub may ask you to authenticate when you clone a private repository or push. Use the authentication method configured for your environment, such as a browser sign-in, a credential manager, or an SSH key. Follow the current course setup guidance if none is configured.

Never paste a password or access token into a notebook, a command saved in the notebook, or a repository URL. URLs and notebook outputs are easy to copy, commit, or expose in logs. If a secret is exposed, stop using it, revoke it through the service that issued it, and ask for help.

## Inspect history from the command line

GitHub's history page and the local log describe the same committed project history. A compact local view is:

```bash
git log --oneline --decorate -n 10
```

This shows recent commit identifiers, messages, and branch labels. Meaningful messages make the log useful for understanding how an analysis developed and for locating a known-good checkpoint.

## A brief collaboration preview

A **branch** is a movable label for one line of development. The default branch is commonly named `main`; on GitHub, it is often the shared baseline for the project. A **merge** combines compatible work from different branches. Branching, merging, and pull requests become more important in team projects, but the checkpoint loop does not change: inspect, stage, commit, and share deliberately.

## Recovery starts from saved copies

A confusing working folder is not the same as lost committed work. When something looks wrong:

1. Stop and run `git status`; read what Git says is modified, staged, untracked, ahead, or behind.
2. Check the history on GitHub or with `git log --oneline`. A useful earlier commit is still recorded.
3. Do not force a push or delete files just to make an error disappear. Ask for help if the message is unclear.
4. If you need an untouched reference copy, clone the GitHub repository into a **different, empty directory**. Keep the confusing folder until you know whether it contains uncommitted work.

The two recovery anchors are the commit history and the GitHub copy of your committed work.

## A small troubleshooting guide

| What you observe | First useful check | Likely next step |
|---|---|---|
| A new file is absent from GitHub | `git status` | Stage the intended file, commit, then push |
| A changed file is absent from the next commit | `git diff --staged` | Stage that file before committing |
| The local branch is behind GitHub | `git status` | Pull after accounting for local edits |
| A push is rejected | Read the entire message | Pull or resolve the named issue; do not force by default |
| A secret appears in a file or URL | Stop immediately | Revoke the secret and get help; deleting one visible line is not enough |


## The workflows to remember

In Colab:

**fork once → open your fork → edit → save a commit → check history**

In a full workspace:

**clone once → pull → edit → status → diff → add → commit → push**

The menus and commands differ, but the underlying habit is the same: make intentional checkpoints, keep them in the correct GitHub repository, and inspect before acting.
